```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# Load the dataset
df = pd.read_csv("Salary_Data.csv")

# Select input and output columns
X = df[["YearsExperience"]]
y = df["Salary"]

# -------------------------------
# 1. Linear Regression
# -------------------------------

linear_model = LinearRegression()
linear_model.fit(X, y)

# Predict salary
linear_predictions = linear_model.predict(X)

print("Linear Regression")
print("Predicted Salaries:")
print(linear_predictions)
print("R2 Score:", r2_score(y, linear_predictions))

# Plot Linear Regression
plt.scatter(X, y, color="blue", label="Actual Data")
plt.plot(X, linear_predictions, color="red", label="Linear Regression")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Linear Regression")
plt.legend()
plt.show()

```

```python
# -------------------------------
# 2. Polynomial Regression
# -------------------------------

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)

# Predict salary
poly_predictions = poly_model.predict(X_poly)

print("\nPolynomial Regression")
print("Predicted Salaries:")
print(poly_predictions)
print("R2 Score:", r2_score(y, poly_predictions))

# Plot Polynomial Regression
plt.scatter(X, y, color="blue", label="Actual Data")
plt.plot(X, poly_predictions, color="green", label="Polynomial Regression")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Polynomial Regression")
plt.legend()
plt.show()
```
